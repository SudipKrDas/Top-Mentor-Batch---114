3.PROCESSING THE DATA

```python
## Checking null values
## Checking number of columns
## Checking number of records

df.isnull().sum()
df.shape
```

```python
X=df['Sorting Time']
X
```

```python
X.shape
```

```python
y=df['Delivery Time']
y
```

4.MODEL THE PROBLEM
    Model Creation - Linear Regression Model
    Machine Learning models - Phython - scikit learn linear regression

```python
## Coverting X and y from series to one dimentional array required for simple linear regression
X=X.reshape(-1,1)
y=y.reshape(-1,1)
print(type(X))
```

```python
##Machine Learning Mpdel
##Linear Regression Model
from sklearn.linear_model import LinearRegression 
```

```python
model=LinearRegression()
```

```python
%time
## Training the ML System (Model)
model_train=model.fit(X,y)
print("Model Training is Completed")
```

```python
## Rules - 
slope=model_train.coef_
print(slope)
```

```python
constant=model_train.intercept_
print(constant)
```

5.VALIDATION AND EXECUTION

```python
## Acuracy of the Model - R Square
r_square=model_train.score(X,y)
print(r_square)
```

```python
##Prediction of y for any value of X
X=[[4,5]]
y=(slope*X)+constant
print(y)
```
