### What are the earliest two films listed in the titles dataframe?

```python
#First Solution
x=cast[['title','year']].sort_values('year').drop_duplicates().head(2)['title'].unique()
```

```python
print(f'The two earliest film in the cast dataframe are: "{x[0]}" and "{x[1]}"')
```

```python
## Second alternative solution using nsmallest:
first2=cast[['title','year']].drop_duplicates("title").nsmallest(2,'year')['title'].unique()
print("The two earliest films in the cast dataframe are: 1.",first2[0],"and 2.",first2[1])
```

```python
###Third Solution using set_index:
x_index=cast.set_index("year").sort_index().drop_duplicates("title").head(2)["title"].unique()
print("The two earliest films in the cast dataframe are: 1.",x_index[0],"and 2.",x_index[1])
```
